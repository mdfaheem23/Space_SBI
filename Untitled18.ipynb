```python
temperature = Isothermal(
    T=1500.0
)

print("Temperature:", 1500, "K")
```

```python
pressure = SimplePressureProfile(
    nlayers=100,
    atm_min_pressure=1e-5,
    atm_max_pressure=1e5
)

print("Number of layers:", 100)
print("Minimum pressure:", 1e-5)
print("Maximum pressure:", 1e5)
```

```python
chemistry = TaurexChemistry(
    fill_gases=["H2", "He"],
    ratio=0.1756
)

print("Background gases: H2 and He")
```

```python
chemistry.addGas(
    ConstantGas(
        molecule_name="H2O",
        mix_ratio=1e-4
    )
)

print("H2O added")

chemistry.addGas(
    ConstantGas(
        molecule_name="CO2",
        mix_ratio=1e-5
    )
)

print("CO2 added")

chemistry.addGas(
    ConstantGas(
        molecule_name="CH4",
        mix_ratio=1e-5
    )
)

print("CH4 added")

chemistry.addGas(
    ConstantGas(
        molecule_name="NH3",
        mix_ratio=1e-5
    )
)

print("NH3 added")
```

```python
star = BlackbodyStar(
    temperature=5778.0,
    radius=1.0
)

print("Star created")
print("Temperature:", 5778, "K")
```

```python
planet = Planet(
    planet_mass=1.0,
    planet_radius=1.0
)

print("Planet created")
```

```python
model = TransmissionModel(
    planet=planet,
    temperature_profile=temperature,
    chemistry=chemistry,
    pressure_profile=pressure,
    star=star
)

print("Transmission model created!")
```

```python
model.add_contribution(
    AbsorptionContribution()
)

print("Molecular absorption added!")
```

```python
model.build()

print("TauREx model successfully built!")
```

```python
wngrid, spectrum, tau, _ = model.model()

print("Simulation completed!")
```

```python
print("Wavenumber shape:", wngrid.shape)
print("Spectrum shape:", spectrum.shape)
print("Optical depth shape:", tau.shape)
```

```python
T = 1500
H20 = 1e-4
CO2 = 1e-5
CH4 = 1e-5
NH3 = 1e-5
```

```python
def simulator(thetha):
  temperature_value = thetha[0]
  h2o = 10** thetha[1]
  co2 = 10** thetha[2]
  ch4 = 10** thetha[3]
  nh3 = 10** thetha[4]

  temperature = Isothermal(
      T = temperature_value
  )

  chemistry = TaurexChemistry(
      fill_gases = ["H2","He"],
      ratio = 0.1756
  )

  chemistry.addGas(
      ConstantGas(
          molecule_name = "H20",
          mix_ratio = h2o
      )
  )
  chemistry.addGas(
      ConstantGas(
          molecule_name = "CO2",
          mix_ratio = co2
      )
  )
  chemistry.addGas(
      ConstantGas(
          molecule_name = "CH4",
          mix_ratio = ch4
      )
  )
  chemistry.addGas(
      ConstantGas(
          molecule_name = "NH3",
          mix_ratio = nh3
      )
  )
  model = TransmissionModel(
      planet =planet,
      temperature_profile=temperature,
      chemistry=chemistry,
      pressure_profile=pressure,
      star=star
  )
  model.add_contribution(
      AbsorptionContribution()
  )
  model.build()

  wngrid, spectrum,lau,_ = model.model()
  return spectrum
```

```python
import torch
from sbi.utils import BoxUniform

low = torch.tensor([
    1000.0,
    -7.0,
    -7.0,
    -7.0,
    -7.0
])

high = torch.tensor([
    2000.0,
    -3.0,
    -3.0,
    -3.0,
    -3.0
])

prior = BoxUniform(
    low=low,
    high=high
)

theta = prior.sample((500,))

print(theta)
print(theta.shape)
```

```python
nums_simulation = 100
thetha = prior.sample((nums_simulation,))
print(thetha.shape)
```

```python
import numpy as np

spectra = []

for i in range(nums_simulation):
    current_thetha = thetha[i].tolist()
    spectrum = simulator(current_thetha)
    spectra.append(spectrum)
    print(f"Simulation {i+1}/{nums_simulation} completed")
spectra = np.array(spectra)

print("Spectra shape:", spectra.shape)
print("Parameters:", theta.shape)
print("Spectra:", spectra.shape)
```
