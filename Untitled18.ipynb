```python
planet = Planet(
    planet_mass=1.0,
    planet_radius=1.0
  )
```

```python
star = BlackbodyStar(
    temperature=5778.0,
    radius=1.0
  )
```

```python
pressure = SimplePressureProfile(
    nlayers=100,
    atm_min_pressure=1e-5,
    atm_max_pressure=1e5
  )
```

```python
def simulator(thetha):
  temperature_value = thetha[0]
  h2o = 10** thetha[1]
  co2 = 10** thetha[2]
  ch4 = 10** thetha[3]
  nh3 = 10** thetha[4]

  temperature = Isothermal(
      T = temperature_value
  )

  print("Planet created")

  print("Star created")
  print("Temperature:", 5778, "K")

  chemistry = TaurexChemistry(
      fill_gases = ["H2","He"],
      ratio = 0.1756
  )

  chemistry.addGas(
      ConstantGas(
          molecule_name = "H2O",
          mix_ratio = h2o
      )
  )
  chemistry.addGas(
      ConstantGas(
          molecule_name = "CO2",
          mix_ratio = co2
      )
  )
  chemistry.addGas(
      ConstantGas(
          molecule_name = "CH4",
          mix_ratio = ch4
      )
  )
  chemistry.addGas(
      ConstantGas(
          molecule_name = "NH3",
          mix_ratio = nh3
      )
  )
  model = TransmissionModel(
      planet =planet,
      temperature_profile=temperature,
      chemistry=chemistry,
      pressure_profile=pressure,
      star=star
  )
  model.add_contribution(
      AbsorptionContribution()
  )
  model.build()

  wngrid, spectrum,lau,_ = model.model()
  return wngrid,spectrum
```

```python
import torch
from sbi.utils import BoxUniform

low = torch.tensor([
    1000.0,
    -7.0,
    -7.0,
    -7.0,
    -7.0
])

high = torch.tensor([
    2000.0,
    -3.0,
    -3.0,
    -3.0,
    -3.0
])

prior = BoxUniform(
    low=low,
    high=high
)
```

```python
nums_simulation = 1000
thetha = prior.sample((nums_simulation,))
print(thetha.shape)
```

```python
import numpy as np

spectra = []

for i in range(nums_simulation):
    current_thetha = thetha[i].tolist()
    wngrid_simulator,spectrum_simulator = simulator(current_thetha)
    spectra.append(spectrum_simulator)
    print(f"Simulation {i+1}/{nums_simulation} completed")
spectra = np.array(spectra)

print("Spectra shape:", spectra.shape)
print("Parameters:", thetha.shape)
print("Spectra:", spectra.shape)
```

```python
wavelength = 10000.0 / wngrid_simulator[::-1]

print("Wavelength points:", len(wavelength))
print("Minimum:", wavelength.min())
print("Maximum:", wavelength.max())
```

```python
import torch
import torch.nn as nn

class SpectrumDetails(nn.Module):

    def __init__(self):
        super().__init__()

        self.fulllayer = nn.Sequential(
            nn.Linear(76744, 256),
            nn.ReLU(),

            nn.Linear(256, 64)
        )

    def forward(self, x):
        return self.fulllayer(x)
```

```python
def simulator_mismatch(theta, planet, pressure, star):

    temperature_value = theta[0]

    h2o = 10**theta[1]
    co2 = 10**theta[2]
    ch4 = 10**theta[3]
    nh3 = 10**theta[4]

    temperature = Isothermal(
        T=temperature_value
    )

    chemistry = TaurexChemistry(
        fill_gases=["H2", "He"],
        ratio=0.30
    )

    chemistry.addGas(
        ConstantGas(
            molecule_name="H2O",
            mix_ratio=h2o
        )
    )

    chemistry.addGas(
        ConstantGas(
            molecule_name="CO2",
            mix_ratio=co2
        )
    )

    chemistry.addGas(
        ConstantGas(
            molecule_name="CH4",
            mix_ratio=ch4
        )
    )

    chemistry.addGas(
        ConstantGas(
            molecule_name="NH3",
            mix_ratio=nh3
        )
    )

    model = TransmissionModel(
        planet=planet,
        temperature_profile=temperature,
        chemistry=chemistry,
        pressure_profile=pressure,
        star=star
    )

    model.add_contribution(
        AbsorptionContribution()
    )

    model.build()

    wngrid, spectrum, tau, _ = model.model()

    return wngrid, spectrum
```

```python
from sbi.inference import NPE_C
from sbi.neural_nets import posterior_nn

ensemble_posteriors = []

for seed in [1, 2, 3]:

    torch.manual_seed(seed)

    embedding_net_i = SpectrumDetails()

    neural_posterior_i = posterior_nn(
        model="maf",
        embedding_net=embedding_net_i
    )

    inference_i = NPE_C(
        prior=prior,
        density_estimator=neural_posterior_i
    )

    density_estimator_i = (
        inference_i
        .append_simulations(
            theta_tensor,
            x_tensor
        )
        .train()
    )

    posterior_i = inference_i.build_posterior(
        density_estimator_i
    )

    ensemble_posteriors.append(
        posterior_i
    )

    print(f"Model {seed} trained")
```

```python
ensemble_means = []

for posterior_i in ensemble_posteriors:

    samples_i = posterior_i.sample(
        (1000,),
        x=x_mismatch_tensor
    )

    mean_i = samples_i.mean(
        dim=0
    )

    ensemble_means.append(mean_i)

ensemble_means = torch.stack(
    ensemble_means
)

print(ensemble_means)
```

```python
ensemble_std = ensemble_means.std(
    dim=0
)

for i, name in enumerate(parameter_names):

    print(
        f"{name}: "
        f"ensemble disagreement = "
        f"{ensemble_std[i].item():.4f}"
    )
```

```python
with torch.no_grad():

    training_embeddings = embedding_net(
        x_tensor
    )

    test_embedding = embedding_net(
        x_mismatch_tensor.unsqueeze(0)
    )

print(training_embeddings.shape)
print(test_embedding.shape)
```

```python
embedding_mean = training_embeddings.mean(
    dim=0
)

centered = (
    training_embeddings - embedding_mean
)

covariance = (
    centered.T @ centered
) / (len(training_embeddings) - 1)

covariance = covariance + (
    1e-5 * torch.eye(
        covariance.shape[0]
    )
)
```

```python
diff = (
    test_embedding.squeeze(0)
    - embedding_mean
)

mahalanobis_distance = torch.sqrt(
    diff @ torch.linalg.solve(
        covariance,
        diff
    )
)

print(
    "Mahalanobis distance:",
    mahalanobis_distance.item()
)
```
